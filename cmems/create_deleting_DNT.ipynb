```python
import glob
from datetime import datetime, timedelta
```

```python
xml_start = """<?xml version='1.0' encoding='UTF-8'?>
<delivery product="SEAICE_ARC_PHY_AUTO_L4_MY_011_025" PushingEntity="SIW-METNO-OSLO-NO" date="20260225T095900">
<dataset DatasetName="cmems_obs-si_arc_phy-siage_my_P1D-m_202603">
"""
xml_end = """</dataset></delivery>
"""
file_entry = '<file FileName="%s" FileType="DT"><KeyWord>Delete</KeyWord></file>\n'


date_start = datetime(1991,9, 15)
dates = [date_start + timedelta(i) for i in range(108)]

date_start = datetime(1992, 1, 1)
dates = [date_start + timedelta(i) for i in range(1353)]
dates[0], dates[-1]
```

```python
xml_content = xml_start
for date in dates:
    filename = date.strftime("%Y/%m/nersc_siage_arctic65_%Y%m%d.nc")
    xml_content += file_entry % filename
xml_content += xml_end
dnt_filename = f'DNT/SEAICE_ARC_PHY_AUTO_L4_MY_011_025_delete_{datetime.now().strftime("%Y%m%d%H%M%S")}.xml'
with open(dnt_filename, 'wt') as f:
    f.write(xml_content)
print(dnt_filename)
```
